```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import DenseNet121
from tensorflow.keras import layers, models

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
IMG_SIZE = 224
BATCH_SIZE = 16

train_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.15,
    rotation_range=15,
    zoom_range=0.1,
    horizontal_flip=True
)

train_generator = train_datagen.flow_from_directory(
    "/content/Training",
    target_size=(IMG_SIZE,IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="training"
)

val_generator = train_datagen.flow_from_directory(
    "/content/Training",
    target_size=(IMG_SIZE,IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="validation"
)

test_datagen = ImageDataGenerator(rescale=1./255)

test_generator = test_datagen.flow_from_directory(
    "/content/Testing",
    target_size=(IMG_SIZE,IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)
```

```python
def build_baseline_model():

    base_model = DenseNet121(
        weights="imagenet",
        include_top=False,
        input_shape=(224,224,3)
    )

    for layer in base_model.layers[:-30]:
        layer.trainable = False

    x = base_model.output

    x = layers.GlobalAveragePooling2D()(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.5)(x)

    outputs = layers.Dense(4, activation="softmax")(x)

    model = models.Model(base_model.input, outputs)

    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-4),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

```python
def channel_attention(x, ratio=16):

    channel = x.shape[-1]

    avg_pool = layers.GlobalAveragePooling2D()(x)
    avg_pool = layers.Reshape((1,1,channel))(avg_pool)

    dense1 = layers.Dense(channel//ratio, activation='relu')(avg_pool)
    dense2 = layers.Dense(channel, activation='sigmoid')(dense1)

    return layers.Multiply()([x, dense2])
```

```python
from tensorflow.keras import layers

def spatial_attention(x):

    avg_pool = layers.Lambda(lambda x: tf.reduce_mean(x, axis=-1, keepdims=True))(x)
    max_pool = layers.Lambda(lambda x: tf.reduce_max(x, axis=-1, keepdims=True))(x)

    concat = layers.Concatenate(axis=-1)([avg_pool, max_pool])

    attention = layers.Conv2D(
        filters=1,
        kernel_size=7,
        padding="same",
        activation="sigmoid"
    )(concat)

    output = layers.Multiply()([x, attention])

    return output
```

```python
def build_dual_model():

    base_model = DenseNet121(
        weights="imagenet",
        include_top=False,
        input_shape=(224,224,3)
    )

    for layer in base_model.layers[:-30]:
        layer.trainable = False

    x = base_model.output

    x = channel_attention(x)
    x = spatial_attention(x)

    x = layers.GlobalAveragePooling2D()(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.5)(x)

    outputs = layers.Dense(4, activation="softmax")(x)

    model = models.Model(base_model.input, outputs)

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=5e-5),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

```python
callbacks = [

    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=3
    )
]
```

```python
callbacks = [

    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=5
    )
]
```

```python
baseline_model = build_baseline_model()

history_baseline = baseline_model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=30,
    callbacks=callbacks
)
```

```python
dual_model = build_dual_model()

history_dual = dual_model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=30,
    callbacks=callbacks
)
```

```python
baseline_loss, baseline_acc = baseline_model.evaluate(test_generator)

dual_loss, dual_acc = dual_model.evaluate(test_generator)

print("Baseline Accuracy:", baseline_acc)
print("Dual Attention Accuracy:", dual_acc)
```

```python
from sklearn.metrics import classification_report

pred = dual_model.predict(test_generator)
pred_classes = np.argmax(pred, axis=1)

print(classification_report(test_generator.classes, pred_classes))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(test_generator.classes, pred_classes)

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, cmap="Blues", fmt="d")

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")

plt.show()
```

```python
models = ["Baseline DenseNet", "Dual Attention DenseNet"]
accuracies = [baseline_acc*100, dual_acc*100]

plt.bar(models, accuracies)

plt.ylabel("Accuracy (%)")
plt.title("Model Accuracy Comparison")

plt.show()
```
